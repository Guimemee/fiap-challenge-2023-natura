Estruturar a modelagem de requisitos funcionais e não-funcionais para um sistema IoT de rastreamento de lotes garantindo transparência de ponta a ponta e precificação justa baseada em carbono evitado.

### Modelagem Matemática Formal
$$\text{Pegada Carbono Evitada} = \sum_{k=1}^{n} (\text{Emissão Padrão}_k - \text{Emissão Real}_k), \quad \text{Índice ESG} \ge 90\%$$


### Passo 1: Identificação dos Atores da Cadeia Extrativista
Mapeamento de 3 cooperativas locais congregando 180 famílias de coletores florestais tradicionais.


### Passo 2: Análise dos Pontos de Atrito e Perdas Pós-Colheita
Identificação de 18% de perda de umidade e proliferação fúngica decorrente do transporte demorado em barcos sem ventilação monitorada.


### Passo 3: Especificação dos Requisitos de Telemetria
Necessidade de registrar temperatura ambiente, umidade relativa, geolocalização e data/hora a cada transbordo de sacaria.


### Passo 4: Cálculo do Impacto Econômico de Certificação
Valorização estimada de 28% no preço de venda da tonelada de sementes certificadas no mercado internacional de cosméticos sustentáveis.


### Passo 5: Conclusão de Escopo
Definição da arquitetura de tags RFID passivas em sacos associadas a leitores LoRaWAN em barcos coletores.


```python
import numpy as np
import matplotlib.pyplot as plt

etapas = ['Coleta Mata', 'Secagem', 'Barco Juruá', 'Cooperativa', 'Fábrica Natura']
perdas_antes = [4.5, 6.2, 5.8, 2.1, 1.2] # %
perdas_depois = [1.2, 1.5, 1.1, 0.4, 0.2] # %
x = np.arange(len(etapas))
w = 0.35
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#0f2d18')
ax.set_facecolor('#07190c')
ax.bar(x - w/2, perdas_antes, width=w, label='Cadeia Tradicional (Perda Total = 19.8%)', color='#ef4444')
ax.bar(x + w/2, perdas_depois, width=w, label='Com Rastreamento IoT (Perda Total = 4.4%)', color='#10b981')
ax.set_xticks(x); ax.set_xticklabels(etapas, fontsize=8, color='#94a3b8')
ax.set_title('Redução de Perdas Pós-Colheita na Cadeia Natura (Médio Juruá)', color='#ffffff', fontsize=10.5, fontweight='bold')
ax.set_ylabel('Taxa de Perda (%)', color='#94a3b8', fontsize=9)
ax.tick_params(colors='#94a3b8')
ax.grid(True, color='#256136', alpha=0.5, ls=':')
ax.legend(facecolor='#174224', edgecolor='#256136', labelcolor='#ffffff', fontsize=8)
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=200, facecolor=fig.get_facecolor())
plt.close()
```
