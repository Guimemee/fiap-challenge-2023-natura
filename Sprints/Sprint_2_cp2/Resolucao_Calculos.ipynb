> **Especificação da Sprint:**
> Dimensionar os nós sensores autônomos operando na frequência LoRaWAN de 915 MHz sob dossel florestal denso com atenuação por folhagem úmida. Projetar o sistema de alimentação com mini painel solar fotovoltaico de 5 V / 200 mA e supercapacitor de 50 F com bateria de suporte Li-ion 18650 para suportar até 5 dias consecutivos de chuva intensa sem insolação direta.

### Modelagem Matemática Formal
$$\text{Atenuação Florestal (Weissberger)} = 1,33 \cdot f^{0,284} \cdot d^{0,588}, \quad P_{\text{painel}} \ge P_{\text{sensor}} \cdot \frac{24}{\text{HSP}}$$


### Passo 1: Modelo de Propagação com Vegetação Densa (ITU-R / Weissberger)
Para f = 915 MHz e profundidade de folhagem d = 80 metros: \text{Atenuação} = 1,33 \times (0,915)^{0,284} \times (80)^{0,588} \approx 1,33 \times 0,975 \times 13,12 \approx 17,01\text{ dB} adicionais.


### Passo 2: Perfil de Consumo Energético do Nó Sensor
Consumo em repouso: 12 \mu\text{A}. Consumo durante transmissão LoRa a +20 dBm (SF10): 110 mA durante 0,42 s a cada 15 minutos. Potência média contínua consumida: P_{med} \approx 1,45\text{ mW}.


### Passo 3: Dimensionamento do Mini Painel Solar Fotovoltaico
Horas de Sol Pleno (HSP) médias na Amazônia em época chuvosa = 2,8 horas/dia. Energia diária demandada = 1,45\text{ mW} \times 24\text{ h} = 34,8\text{ mWh}. Geração solar mínima necessária = 34,8 / 2,8 \approx 12,4\text{ mW}.


### Passo 4: Capacidade do Supercapacitor e Bateria Li-ion
Supercapacitor de 50 F fornece pico de corrente durante rajadas de transmissão sem desgastar os ciclos químicos da bateria de 2.600 mAh.


### Passo 5: Conclusão de Autonomia Perpétua
A bateria de 2.600 mAh garante 78 dias de operação contínua mesmo em escuridão completa, superando com margem a exigência de 5 dias.


```python
import numpy as np
import matplotlib.pyplot as plt

dist_floresta = np.linspace(10, 200, 100) # metros dentro da mata
att_floresta = 1.33 * (0.915**0.284) * (dist_floresta**0.588)
prx_com_mata = 14 + 2 + 3 - (20*np.log10(dist_floresta/1000) + 20*np.log10(915) + 32.44) - att_floresta
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#0f2d18')
ax.set_facecolor('#07190c')
ax.plot(dist_floresta, prx_com_mata, color='#86efac', lw=2.5, label='Potência com Atenuação de Folhagem (dBm)')
ax.axhline(-137, color='#ef4444', ls='--', lw=1.5, label='Sensibilidade LoRa SF12 (-137 dBm)')
ax.set_title('Atenuação de Rádio LoRa sob Dossel Florestal Amazônico', color='#ffffff', fontsize=10.5, fontweight='bold')
ax.set_xlabel('Distância através da Mata Fechada (Metros)', color='#94a3b8', fontsize=9)
ax.set_ylabel('Nível de Sinal Recebido (dBm)', color='#94a3b8', fontsize=9)
ax.tick_params(colors='#94a3b8')
ax.grid(True, color='#256136', alpha=0.5, ls=':')
ax.legend(facecolor='#174224', edgecolor='#256136', labelcolor='#ffffff', fontsize=8)
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=200, facecolor=fig.get_facecolor())
plt.close()
```
